```python
import pandas as pd
import matplotlib.pyplot as plt

%matplotlib inline

data = pd.read_csv(
    r'C:\Users\user\Desktop\data science\data_science_salaries.csv'
)

# Boxplot by company size
data.boxplot(column='salary_in_usd', by='company_size')
plt.title('Salary Distribution by Company Size')
plt.suptitle('')
plt.xlabel('Company Size')
plt.ylabel('Salary in USD')
plt.savefig('salary_by_company_size.png', bbox_inches='tight')
plt.show()

# Boxplot by experience level
data.boxplot(column='salary_in_usd', by='experience_level')
plt.title('Salary Distribution by Experience Level')
plt.suptitle('')
plt.xlabel('Experience Level')
plt.ylabel('Salary in USD')
plt.savefig('salary_by_experience_level.png', bbox_inches='tight')
plt.show()

# Bar chart
plot_data = data.groupby(
    ['company_size', 'experience_level']
)['salary_in_usd'].mean().unstack()

plot_data = plot_data.reindex(index=['S', 'M', 'L'])
plot_data = plot_data[['EN', 'MI', 'SE', 'EX']]

plot_data.plot.bar()
plt.ylabel('USD Salary')
plt.title('Average Salary by Company Size and Experience Level')
plt.savefig('avg_salary_company_experience.png', bbox_inches='tight')
plt.show()
```
